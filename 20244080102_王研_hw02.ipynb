```python
# 1 判断回文 is_palindrome
def is_palindrome(s: str) -> bool:
    # 去掉空格，全部转为小写
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]


# 2 判断字母顺序 is_abecedarian
def is_abecedarian(s: str) -> bool:
    # 去掉空格，转小写
    clean = s.replace(" ", "").lower()
    for i in range(len(clean)-1):
        if clean[i] > clean[i+1]:
            return False
    return True


#3 删除元音 remove_vowels
def remove_vowels(s: str) -> str:
    vowels = {"a", "e", "i", "o", "u"}
    res = []
    for ch in s:
        if ch.lower() not in vowels:
            res.append(ch)
    return "".join(res)


# 测试样例
print("===is_palindrome测试===")
print(is_palindrome("T A C O cat"))
print(is_palindrome("level"))
print(is_palindrome("hello"))

print("\n===is_abecedarian测试===")
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("cat"))

print("\n===remove_vowels测试===")
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

```python
#1 list_reverse，输入不是列表抛异常
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]


#2 is_sorted 只遍历一次，判断升序
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True


#测试
print("===list_reverse测试===")
print(list_reverse([1,2,3]))

print("\n===is_sorted测试===")
print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2,4]))
print(is_sorted((2,4,5)))
```

```python
#1 向量标量乘法 vec_scalar_mult
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组向量")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或者浮点数")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元组元素必须是数字")
    return tuple(x * s for x in t)


#2 向量内积 vec_inner_product
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("输入必须是元组向量")
    if len(x) != len(y):
        raise ValueError("两个向量维度（长度）必须相等")
    for xi in x:
        if not isinstance(xi, (int, float)):
            raise TypeError("向量元素必须是数字")
    for yi in y:
        if not isinstance(yi, (int, float)):
            raise TypeError("向量元素必须是数字")
    total = 0.0
    for xi, yi in zip(x, y):
        total += xi * yi
    return total


#3 check_valid_mx 检查矩阵是否合法（元组的元组）
def check_valid_mx(mx):
    #外层必须是元组
    if not isinstance(mx, tuple):
        return False
    #每一行必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        #所有行长度相同
        if len(row) != row_len:
            return False
        #每个元素必须int/float
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


#4 mx_vec_mult 矩阵乘向量
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise TypeError("矩阵格式不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    m = len(mx)
    if m ==0:
        return ()
    n_cols = len(mx[0])
    if len(vec) != n_cols:
        raise ValueError(f"矩阵列数{n_cols}和向量长度{len(vec)}不匹配")
    out = []
    for row in mx:
        out.append(vec_inner_product(row, vec))
    return tuple(out)


#测试
print("===vec_scalar_mult===")
print(vec_scalar_mult((1,2,3), 2))

print("\n===vec_inner_product===")
print(vec_inner_product((1,2),(3,4)))

print("\n===check_valid_mx===")
mat = ((1,2),(3,4))
print(check_valid_mx(mat))

print("\n===mx_vec_mult===")
print(mx_vec_mult(mat, (5,6)))
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        # dim维度校验
        if not isinstance(dim, int):
            raise TypeError("维度dim必须是整数")
        if dim < 0:
            raise ValueError("维度不能是负数")

        # 没有给elements，默认全0
        if elements is None:
            elements = [0]*dim
        else:
            elements = list(elements)

        if len(elements) != dim:
            raise ValueError(f"维度{dim}与元素数量{len(elements)}不匹配")

        for e in elements:
            if not isinstance(e, (int, float)):
                raise TypeError("向量元素只能是整数或浮点数")

        #实例属性
        self.dim = dim
        self.elements = elements

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector实例比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector实例比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector实例比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的Vector")
        return self.elements > other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot参数必须是Vector对象")
        if self.dim != other.dim:
            raise ValueError("内积要求两个向量维度相等")
        s = 0.0
        for a,b in zip(self.elements, other.elements):
            s += a*b
        return s

    # v * other
    def __mul__(self, other):
        #向量 * 标量
        if isinstance(other, (int, float)):
            new_elems = [x*other for x in self.elements]
            return Vector(self.dim, new_elems)
        #向量 * 向量：逐元素相乘哈达玛积
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量相乘维度必须一致")
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("乘法只支持标量(int/float)或者Vector实例")

    # 标量 * v  2*v  左乘
    def __rmul__(self, scalar):
        return self * scalar

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"


# =========测试Vector类=========
v1 = Vector(2, [2,4])
v2 = Vector(2, [2,5])
print("v1=",v1)
print("v2=",v2)
print("v1 < v2 ?", v1 < v2)
print("v1 == v2 ?", v1 == v2)

#标量乘法
print("v1*3 =", v1*3)
print("3*v1 =",3*v1)

#向量逐元素相乘
v3 = Vector(2,(2,3))
v4 = Vector(2,(4,5))
print("v3*v4 =", v3*v4)

#内积dot
print("v3.dot(v4) =", v3.dot(v4))
```
