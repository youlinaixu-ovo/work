```python
#实验 1 NumPy 数组、数据类型与索引
#任务 1：int/float/bool 数组，对比 python list
# 创建三种数组
arr_int = np.array([1,2,3])
arr_float = np.array([1.1,2.2,3.3])
arr_bool = np.array([True, False, True])

for arr,name in zip([arr_int,arr_float,arr_bool],["int数组","float数组","bool数组"]):
    print(f"==== {name} ====")
    print(f"dtype={arr.dtype}, shape={arr.shape}, ndim={arr.ndim}")
    print(arr)
    
#任务 2 arange (12) reshape
a = np.arange(12)
print("原始一维数组：", a)
a1 = a.reshape(3,4)
a2 = a.reshape(2,3,2)
print("\nreshape(3,4):\n",a1)
print("\nreshape(2,3,2):\n",a2)

#任务 3 对 (3,4) 数组索引
arr = np.arange(12).reshape(3,4)
print("原数组：\n",arr)

row2 = arr[1,:]   # 第2行，下标从0开始
col1_3 = arr[:,0:3] #第1~3列
gt6 = arr[arr>6]    #大于6的元素

print("\n第2行：", row2, "shape", row2.shape)
print("\n1‑3列：\n", col1_3, "shape", col1_3.shape)
print("\n大于6元素：", gt6, "shape", gt6.shape)

#任务 4 布尔或、且
x = np.arange(10)
#小于2 或者大于7
res_or = x[(x<2) | (x>7)]
#大于3并且小于7
res_and = x[(x>3) & (x<7)]

#等价用函数写法
res_or2 = x[np.logical_or(x<2, x>7)]
res_and2 = x[np.logical_and(x>3, x<7)]

print("小于2或大于7：", res_or)
print("大于3且小于7：", res_and)

#任务 5 逐元素乘法 vs 矩阵乘法
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

elem1 = A * B
elem2 = np.multiply(A,B)

mat1 = A @ B
mat2 = np.matmul(A,B)

print("A*B逐元素:\n",elem1)
print("np.multiply:\n",elem2)
print("@矩阵乘法:\n",mat1)
print("np.matmul:\n",mat2)
```

```python
#实验 2：随机抽样、统计、缺失值
np.random.seed(42) #固定随机种子！

#1 生成1000个成绩，均值70 std10，限制0‑100
scores = np.random.normal(loc=70, scale=10, size=1000)
scores = np.clip(scores,0,100)

print(f"样本数:{len(scores)}")
print(f"min:{scores.min():.2f}, max:{scores.max():.2f}")
print(f"均值:{scores.mean():.2f},标准差:{scores.std():.2f}")
print(f"极差:{scores.max()-scores.min():.2f}")

#2 不放回抽样
pool = np.arange(1,11)
sample_uniform = np.random.choice(pool,size=5,replace=False)
sample_weighted = np.random.choice(pool,size=5,replace=False,p=pool/np.sum(pool))
print("\n均匀抽样", sample_uniform)
print("按大小加权抽样", sample_weighted)

#3原地打乱 vs 返回新数组
ori = np.array([1,2,3,4,5])
arr_shuffle_new = np.random.permutation(ori) #返回新，原数组不变
print("permutation后原数组不变 ori =", ori)

ori2 = np.array([1,2,3,4,5])
np.random.shuffle(ori2) #原地修改
print("shuffle原地打乱 ori2 =", ori2)

#4 设置nan缺失值
scores_nan = scores.copy()
scores_nan[[9,19,29]] = np.nan #第10、20、30，下标从0开始
count_nan = np.sum(np.isnan(scores_nan))
mean_normal = np.mean(scores_nan)   #遇到nan直接输出nan
mean_ignore_nan = np.nanmean(scores_nan) #跳过nan
print(f"\n缺失值数量 {count_nan}")
print(f"普通np.mean={mean_normal}")
print(f"np.nanmean忽略缺失 {mean_ignore_nan:.2f}")

#5 构造内存表格，给实验4用，不写磁盘
df_exp2 = pd.DataFrame({"score":scores_nan, "idx":np.arange(len(scores_nan))})
print("\nexp2内存表格预览")
print(df_exp2.head())
```

```python
#实验 3 SciPy + Matplotlib
from scipy import linalg, optimize, stats

##任务1 解方程组：3x+y=9; x+2y=8
A_mat = np.array([[3, 1],[1, 2]])
b_vec = np.array([9,8])
sol = linalg.solve(A_mat,b_vec)
print("方程组解 x,y =", sol)
#代入验证
print("代入验证 A@sol =", A_mat @ sol)

##任务2 求x²‑2=0根
def f(x):
    return x**2 -2

root_pos = optimize.root(f, x0=1) #初值1求正根
root_neg = optimize.root(f, x0=-1) #初值‑1求负根
print("\n正根求解成功?", root_pos.success, "结果",root_pos.x)
print("负根求解成功?", root_neg.success, "结果",root_neg.x)
print("f(正根)",f(root_pos.x), "f(负根)", f(root_neg.x))

##任务3最小化 f=x²+x+2
def fun_min(x):
    return x**2 + x + 2
res_min = optimize.minimize(fun_min, x0=0)
print("\n极小值点x:", res_min.x, "函数值f(x)", res_min.fun)
#解析解 x=-0.5，最小值1.75

##任务4独立样本t检验
np.random.seed(10)
g1 = np.random.normal(0,1,size=100)
g2 = np.random.normal(1,1,size=100)
t_res = stats.ttest_ind(g1,g2)
print("\nt‑statistic:",t_res.statistic, "p‑value:", t_res.pvalue)
alpha =0.05
if t_res.pvalue < alpha:
    print("p<0.05：拒绝原假设，两组均值存在显著差异")
else:
    print("p≥0.05，不能拒绝原假设")

##任务5绘图
plt.figure(figsize=(12,4))
#子图1：成绩直方图
plt.subplot(1,3,1)
plt.hist(scores, bins=20)
plt.title("成绩直方图")
plt.xlabel("分数")
plt.ylabel("频数")

#子图2箱线图
plt.subplot(1,3,2)
plt.boxplot(scores)
plt.title("成绩箱线图")

#子图3 y=x, x²,x³ 0~5
plt.subplot(1,3,3)
x_line = np.linspace(0,5,100)
plt.plot(x_line, x_line, label="$y=x$")
plt.plot(x_line, x_line**2, label="$y=x^2$")
plt.plot(x_line, x_line**3, label="$y=x^3$")
plt.legend()
plt.title("函数图像")
plt.xlabel("x")
plt.ylabel("y")
plt.tight_layout()
plt.show()
```

```python
#实验 4 pandas DataFrame
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}
df = pd.DataFrame(records)
df = df.set_index("Name") #行索引设置为姓名

#1新增总分、平均分
df["Total"] = df["Math"]+df["English"]+df["Python"]
df["Average"] = df["Total"]/3

print("行索引：",df.index.tolist())
print("列名：",df.columns.tolist())
print("\n前3行")
print(df.head(3))
print("\n描述统计")
print(df.describe())

#2 loc iloc取Carol；筛选平均分≥80
print("\n---loc取Carol")
print(df.loc["Carol"])
print("\n---iloc取Carol(Carol是第2行下标=1)")
print(df.iloc[1])

print("\n平均分≥80学生")
print(df[df["Average"]>=80])

#3 axis=0列均值；axis=1行均值
print("\n各科列均值(axis=0)")
print(df[["Math","English","Python"]].mean(axis=0))
print("\n每个学生行均值(axis=1)")
print(df[["Math","English","Python"]].mean(axis=1))

#4 apply求最高分；agg min mean max；rank排名
df["max_subject"] = df[["Math","English","Python"]].apply(max, axis=1)
print("\n每人单科最高分：")
print(df["max_subject"])

print("\n每科agg min mean max")
print(df[["Math","English","Python"]].agg(["min","mean","max"]))

df["math_rank"] = df["Math"].rank(ascending=False)
print("\n数学科目排名")
print(df[["Math","math_rank"]])

#5 groupby分组 + transform
group_obj = df.groupby("Group")
print("\n分组人数：")
print(group_obj.size())
print("\n分组三科平均分")
print(group_obj[["Math","English","Python"]].mean())

#transform添加本组Python平均分到每行
df["group_python_mean"] = group_obj["Python"].transform("mean")
print("\n增加分组Python平均分列")
print(df)

#6 filter筛选组：组Python平均分>=80
group_filtered = group_obj.filter(lambda g: g["Python"].mean()>=80)
print("\nfilter过滤后组")
print(group_filtered)

#7协方差、相关系数矩阵
score_cols = ["Math","English","Python"]
cov_matrix = df[score_cols].cov()
corr_matrix = df[score_cols].corr()
print("\n协方差矩阵")
print(cov_matrix)
print("\n皮尔逊相关系数矩阵")
print(corr_matrix)
```
