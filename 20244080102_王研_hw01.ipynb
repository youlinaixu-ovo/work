```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

#测试三组
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))

#测试负数
print(gcd(-12, -8))
```

```python
import math

def euler_limit(n):
    return (1 + 1/n)**n if n !=0 else 1.0


def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1 #阶乘，0! = 1
    for k in range(n):
        if k>0:
            fact = fact * k
        total += 1.0 / fact
    return total


def euler_approx(eps):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - true_e) >= eps:
        if k>0:
            fact = fact * k
        total += 1.0/fact
        k +=1
    return total


def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(val)


#测试
print("euler_limit(1000):", euler_limit(1000))
print("euler_infinite_sum(4):", euler_infinite_sum(4)) #≈2.666666
print("euler_approx(0.001):", euler_approx(0.001))
print("\n---table---")
print_euler_sum_table(5)
```

```python
def is_power_of_2(n):
    # n是正整数
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True


def is_power(b, n):
    while n >1:
        if n % b != 0:
            return False
        n = n // b
    return True


#测试
print(is_power_of_2(8)) #True 2^3
print(is_power_of_2(9)) #False
print(is_power(3,81)) #True 3^4
print(is_power(5,10)) #False
```
