```python
#实验二
# 1.分数等级函数
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须位于0~100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

print(score_level(96))
print(score_level(82))
print(score_level(70))
print(score_level(45))

#2.计算1~100的和
sum_for = 0
for i in range(1, 101):
    sum_for += i
print("for循环1‑100和：", sum_for)

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i +=1
print("while循环1‑100和：", sum_while)

#2.3 迭代、递归阶乘
def fact_iter(n):
    res=1
    for i in range(1,n+1):
        res *=i
    return res

def fact_recur(n):
    if n ==0 or n ==1:
        return 1
    return n * fact_recur(n-1)

print("\n阶乘测试：")
for n in [0,1,5,10]:
    print(f"n={n},迭代:{fact_iter(n)},递归:{fact_recur(n)}")

#2.4 summary 返回最小、最大、平均值
def summary(values):
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)
val = [88,72,95,66]
min_v,max_v,avg_v = summary(val)
print("\nsummary函数：", min_v,max_v,avg_v)

#2.5 可变参数sum_all
def sum_all(*args):
    return sum(args)
print("\nsum_all测试：")
print(sum_all())
print(sum_all(10))
print(sum_all(1,2,3,4))
```

```python
names = ["张三","李四","王五","赵六"]
scores = [85,59,92,76]

#1 zip构建字典
stu_dict = dict(zip(names, scores))
print("姓名成绩字典：", stu_dict)
for k,v in stu_dict.items():
    print(k, v)

#2 enumerate从序号1打印
print("\n带序号遍历：")
for idx, (name,sc) in enumerate(zip(names,scores), start=1):
    print(idx, name, sc)

#3 列表推导式
pass_names = [name for name,sc in zip(names,scores) if sc >=60]
sq_scores = [s**2 for s in scores]
print("\n及格学生：", pass_names)
print("成绩平方列表：", sq_scores)

#4 统计
total = len(scores)
avg = sum(scores)/total
maxs = max(scores)
mins = min(scores)
pass_cnt = sum(1 for s in scores if s>=60)
pass_rate = pass_cnt / total
print(f"\n总人数:{total},平均分:{avg:.2f},最高:{maxs},最低:{mins},及格率:{pass_rate:.2%}")

#5 浅拷贝对比
alias = scores
copy_sc = scores.copy()
scores[0] = 999
print("\n直接赋值alias：", alias)
print(".copy副本：", copy_sc)

#6 元组不可修改
t = (10,20,30)
try:
    t[0] = 99
except TypeError as e:
    print("\n元组修改报错：", e)

#zip长短不一
list1 = [1,2]
list2 = [10,20,30]
print("\nzip长短序列：", list(zip(list1,list2)))
```

```python
import os, pickle
from student_utils import parse_student

# 路径
data_dir = "exp01_data"
file_path = os.path.join(data_dir, "students.txt")

valid_records = []
error_log = []

#读取文件
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()[1:] #跳过表头 name,score
    for line_no, line in enumerate(lines, start=2):
        try:
            res = parse_student(line)
            if res:
                valid_records.append(res)
        except Exception as e:
            error_log.append(f"行{line_no}: {line.strip()}, 错误信息：{e}")

print("✅合法学生记录：")
for rec in valid_records:
    print(rec)

print("\n❌错误行日志：")
for err in error_log:
    print(err)

#统计
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_score = sum(i["score"] for i in valid_records)/valid_num
    pass_count = sum(1 for i in valid_records if i["score"]>=60)
    pass_rate = pass_count / valid_num
else:
    avg_score = 0
    pass_rate=0

#写入report.txt
report_path = os.path.join(data_dir, "report.txt")
with open(report_path,"w",encoding="utf-8") as f:
    f.write(f"有效记录:{valid_num}\n无效记录:{invalid_num}\n平均分:{avg_score:.2f}\n及格率:{pass_rate:.2%}")
print(f"\n📄统计结果已经写入 {report_path}")

#pickle序列化保存
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

#读取pickle验证
with open(pkl_path, "rb") as f:
    load_data = pickle.load(f)
print("\n📦 pickle读取验证：", load_data)

#测试捕获文件不存在异常
try:
    with open(os.path.join(data_dir,"no_file.txt"),"r",encoding="utf-8") as f:
        pass
except FileNotFoundError:
    print("\n⚠️捕获异常：文件不存在")
```

```python
import copy

#父类Student
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >=60
    def __str__(self):
        return f"Student(name:{self.name}, score:{self.score})"

#子类GraduateStudent
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name:{self.name}, score:{self.score}, research:{self.research_topic})"

#实例化
s1 = Student("王研", 82)
print(s1, "是否及格：", s1.is_passed())
g1 = GraduateStudent("小李", 75, "数据分析")
print(g1, "是否及格：", g1.is_passed())

#深浅拷贝演示
origin = {"data":[1,2,3]}
alias = origin
shallow = copy.copy(origin)
deep = copy.deepcopy(origin)

origin["data"][0] = 999
print("\n直接赋值alias:", alias)
print("浅拷贝shallow:", shallow)
print("深拷贝deep:", deep)
```
